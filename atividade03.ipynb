#operadores de comparação


```
'''
operadores de comparação
comparam dois valores e sempre devolvem true ou false
'''

'''
igual a: == ex: 2 == 2, resultado True
diferente de: != ex: 2 != 3, resultado True
maior que: > ex: 2 > 3, resultado False
menor que: < ex: 2 < 3, resultado True
maior ou igual a: >= ex: 2 >= 3, resultado False
menor ou igual a: <= ex: 2 <= 3, resultado True



'''
```

```
idade = 37
print(idade >=18)

print(idade != 18)

print(10 == 10.0)
print("10"==10)
```

```
#comparação de textos

print('Ana'=='ana')
print('Banana' > 'Abacaxi')
print("Z" < 'a')
```

```
#comparação encadeada

nota = 8
print(0 <= nota <=7)

```

```
#operadores lógicos: and, or, not

'''
E (And)
- Será verdadeiro quando todas as condições sejam verdadeiras
- ex: entra quem tem ingresso E é maior de idade

ou (OR)
- Será verdadeiro quando uma condições for verdadeira
- Ex: entra quem tem ingresso ou é maior de idade

não(not)
-  inverte o valor
- ex: se não estiver chovendo, vá a praia
'''
```

```
idade = 20
tem_ingresso = False

print(idade >= 18 and tem_ingresso)
print(idade >= 18 or tem_ingresso)
print(not tem_ingresso)
```

#estruturas condicionais(IF)

```
temperatura = 32

if temperatura > 30:
  print('Muito quente')
  print('fim do programa')
```

```
#if/else
idade = int(input('idade: '))

if idade >= 18:
  print('maior de idade')
else:
  print('menor de idade')
```

```
n1 = float(input('digite a primeira nota: '))
n2 = float(input('digite a segunda nota: '))
n3 = float(input('digite a terceira nota: '))
n4 = float(input('digite a quarta nota: '))

media = (n1+n2+n3+n4)/4

if media >= 7:
  print("aprovado")
else:
  print("não foi aprovado")
```
